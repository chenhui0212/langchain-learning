## 模型调用


#### 加载环境

```python
from dotenv import load_dotenv

# 读取 .env 配置
load_dotenv(override=True)
```

#### 初始化模型

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model='deepseek:deepseek-flash'
)
```

#### 普通调用

```python
resp = llm.invoke("请用一句话介绍你自己")
resp.content
```

#### 流式调用

```python
for chunk in llm.stream("帮我解释一下什么是人工智能"):
    print(chunk.text, end="", flush=True)
```

#### 批量调用

```python
messages = [
    "你好，你是谁？",
    "2 + 3 * 4 = ？",
    "中国首都在哪里？"
]

responses = llm.batch(messages)
for resp in responses:
    print(resp.content)
```

#### 调用参数

允许在调用模型时，动态地分配和控制模型的行为。

```python
from langchain_core.runnables import RunnableConfig
from langchain.chat_models import init_chat_model

llm2 = init_chat_model(
    model='deepseek:deepseek-flash',
    # 指定可调整参数
    configurable_fields=("model", "model_provider", "temperature")
)

config: RunnableConfig = {
    "run_name": "test",
    "metadata": {"userId": "123"},
    # 覆盖模型参数
    "configurable": {
        "model": "deepseek:deepseek-v4-pro",
        "temperature": 1.5
    }
}

resp = llm2.invoke("请用一句话介绍你自己", config=config)
print(resp.model_dump_json(indent=2))
```
