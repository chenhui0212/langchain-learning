#### model_kwargs

补充 LangChain 没有显式声明、但底层 SDK 接受的参数，展开后传给 SDK

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model='deepseek:deepseek-flash',
    model_kwargs={'tools': [
        {
            "type": "function",
            "function": {
                "name": "get_weather",
                "description": "Get weather of a location, the user should supply a location first.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "location": {
                            "type": "string",
                            "description": "The city and state, e.g. San Francisco, CA"
                        }
                    },
                    "required": ["location"]
                }
            }
        }
    ]}
)

resp = llm.invoke("北京今天天气如何")
# 返回包含 tool_calls 信息
print(resp.model_dump_json(indent=2))
```

 #### extr_body

传递供应商扩展的请求体字段，即使 SDK 没有对应的具名参数

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model='deepseek:deepseek-flash',
    # 默认思考是开启的
    extra_body={"thinking": {"type": "disabled"}}
)

resp = llm.invoke("请用一句话介绍你自己")
# 返回不在包含 reasoning_content 信息
print(resp.model_dump_json(indent=2))
```
