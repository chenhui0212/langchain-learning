## 本地模型

#### Ollama

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model='deepseek-r1:1.5b',
    # 默认接口可不指定
    base_url='http://localhost:11434'
)

resp = llm.invoke("请用一句话介绍你自己")
resp.content
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model='deepseek-r1:1.5b',
    model_provider='ollama'
)

resp = llm.invoke("请用一句话介绍你自己")
resp.content
```
