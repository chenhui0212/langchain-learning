#### 简单测试

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model='deepseek:deepseek-flash'
)

llm.invoke("你好")
```

#### 额外参数

```python
from langchain_core.runnables import RunnableConfig
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model='deepseek:deepseek-flash',
)

config: RunnableConfig = {
    "run_name": "tracing-test",
    "tags": ["my_tag1", "my_test"],
    "metadata": {
        "user_id": "123",
        "session_id": "sess_123",
    }
}

llm.invoke("你好呀", config=config)
```
