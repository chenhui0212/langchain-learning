## 结构化输出

#### 加载环境并初始化

```python
from dotenv import load_dotenv

# 读取 .env 配置
load_dotenv(override=True)
```

```python
from langchain.chat_models import init_chat_model
from typing_extensions import TypedDict, Annotated

llm = init_chat_model(
    model='deepseek:deepseek-flash',
    # 需关闭思考模型
    extra_body={
        "thinking": {"type": "disabled"}
    }
)
```

#### TypedDict

```python
class Movie(TypedDict):
    """电影信息"""
    title: Annotated[str, "电影名"]
    year: Annotated[int, "发行年份"]
    director: Annotated[str, "导演"]
    rating: Annotated[float, "电影评分，满分十分"]

structured_llm = llm.with_structured_output(Movie)
resp = structured_llm.invoke("给出电影盗梦空间的信息")
print(resp)
```

##### 嵌套结构

```python
class Address(TypedDict):
    """地址信息"""
    city: Annotated[str, "城市"]
    district: Annotated[str, "区域"]

class Company(TypedDict):
    """公司信息"""
    name: Annotated[str, "公司名称"]
    address: Annotated[Address, "地址"]

structured_llm = llm.with_structured_output(Company)
resp = structured_llm.invoke("阿里巴巴在杭州的滨江区")
print(resp)
```
