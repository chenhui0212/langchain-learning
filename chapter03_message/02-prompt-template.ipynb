## 提示词模板

#### 加载环境并初始化

```python
from dotenv import load_dotenv

# 读取 .env 配置
load_dotenv(override=True)
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model='deepseek:deepseek-flash'
)
```

#### ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# 提示词模版
prompt_template = ChatPromptTemplate([
    ("system", "你是一个友好的 AI 助手，你的名字叫{name}"),
    ("human", "{user_input}")
])

# 三种不同的方式，返回值也不同
prompt_value = prompt_template.invoke({"name": "小智", "user_input": "1 + 1 = ?"})
# prompt_value = prompt_template.format(name="小智", user_input="1 + 1 = ?")
# prompt_value = prompt_template.format_messages(name="小智", user_input="1 + 1 = ?")
print(prompt_value)
print(type(prompt_value))

resp = llm.invoke(prompt_value)
print(resp.text)
```

#### 预填充

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate([
    ("system", "你是{role}，目标用户是{audience}"),
    ("human", "{task}")
])

final_template = prompt_template.partial(role="导游", audience="游客")

prompt_value = final_template.invoke({"task": "介绍一下北京的故宫"})
print(prompt_value)
```

#### 占位

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.prompts.chat import MessagesPlaceholder

prompt_template = ChatPromptTemplate([
    ("system", "你是一个 AI 助手"),
    # 两种写法
    # ("placeholder", "{history}"),
    MessagesPlaceholder("history"),
    ("human", "{question}")
])

prompt_value = prompt_template.invoke({
    "history": [
        ("human", "你好，请问明天天气如何？}"),
        ("ai", "明天天气晴朗"),
    ],
    "question": "后天天气如何"
})
print(prompt_value)
```
